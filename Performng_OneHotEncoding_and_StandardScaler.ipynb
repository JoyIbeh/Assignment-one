### Handling Missing Values


```python
# Filling age with missing values
customer_data['Age']= customer_data['Age'].fillna(customer_data['Age'].median())
```

```python
customer_data.head(20)
```

```python
#fill common missing values in column with the mode

def fill_missing_values(df,column_name):
  df[column_name]= df[column_name].fillna(df[column_name].mode()[0])
  return

column_to_fill = ['Gender','City','Education', 'MaritalStatus', 'Purchased']
for column in column_to_fill:
  fill_missing_values(customer_data,column)
```

```python

customer_data.head(20)
```

```python
customer_data['City'].unique()
```

### Dropping rows NAN rows in Income and Loyalty score


```python
customer_data = customer_data.dropna(subset=['Income','LoyaltyScore'])
```

```python
customer_data.info()
```

```python
# convert the age and Income column to integer  except their are babies, in this case babies do not buy s=product for themselves

column_to_convert = ['Age','Income']
for column in column_to_convert:
  customer_data[column] = customer_data[column].astype(int)
```

# Encode Categorical Data

```python
# creating the object of the encoder
le = LabelEncoder()
```

```python
# calling the encoder
column_to_encode = ['Gender','Purchased']
for column in column_to_encode:
  customer_data[column] = le.fit_transform(customer_data[column])
```

```python
customer_data.head()
```

```python
customer_data['MaritalStatus'].unique()
```

### using oneHot Encoder for multiple columns

```python
columns_to_transform = ['City','Education','MaritalStatus']
column_transformer = ColumnTransformer(transformers=[('encoder',OneHotEncoder(handle_unknown='ignore'),columns_to_transform)],remainder='passthrough')
```

```python
encoded_array = column_transformer.fit_transform(customer_data)
```

```python
customer_data = pd.DataFrame(
    encoded_array,
    columns=column_transformer.get_feature_names_out()
)
display(customer_data.head())
```
